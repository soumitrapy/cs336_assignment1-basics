```python
def train_one_epoch(model, traindl, optimizer, loss_fn):
    model.train()
    total_loss = 0.0
    for x, y in traindl:
        optimizer.zero_grad()
        logits = model(x)
        loss = loss_fn(logits.view(-1, logits.size(-1)), y.view(-1))
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(traindl)
```
