```python
# a.
vocab_size=50257
context_len=1024
d_model=1600
num_layers=48
num_heads=25
d_ff=4288
rope_theta=10000.0
#lm = TransformerLM(vocab_size=vocab_size, context_len=context_len, d_model=d_model, num_layers=num_layers, num_heads=num_heads, d_ff=d_ff, rope_theta=rope_theta)
total_params = 2*vocab_size*d_model + num_layers*(4*d_model*d_model+3*d_model*d_ff+2*d_model)+ d_model
print(f"Total parameters: {total_params}, total parameters in GB (in fp32): {total_params*4/(1e9)}")
```

```python
# c.
vocab_size=50257
context_len=1024
d_model=1600
num_layers=48
num_heads=25
d_ff=4288
rope_theta=10000.0

seq_len = context_len
d_k = d_model // num_heads

qkvo_proj_count = 8*seq_len*d_model*(num_heads*d_k)
scaled_dot_count = 4*(seq_len**2)*(num_heads*d_k) + 3*(seq_len**2)*num_heads
rope_count = 6*seq_len*(num_heads*d_k) 
MultiheadAttention_count = qkvo_proj_count + scaled_dot_count + rope_count
ffn_count = 6*seq_len*d_model*d_ff + 6*seq_len*d_ff
TransformerBlock_count = MultiheadAttention_count + ffn_count + 2*seq_len*d_model 
TransformerLM_count = num_layers*TransformerBlock_count + 3*seq_len*d_model + 2*seq_len*d_model*vocab_size
print(f"Total FLOPS: {TransformerLM_count}, ffn FLOPS: {num_layers*ffn_count}({num_layers*ffn_count/TransformerLM_count*100:.2f}%), multihead attention FLOPS: {num_layers*MultiheadAttention_count}({num_layers*MultiheadAttention_count/TransformerLM_count*100:.2f}%), scaled dot product FLOPS: {num_layers*scaled_dot_count}({num_layers*scaled_dot_count/TransformerLM_count*100:.2f}%), qkvo projection FLOPS: {num_layers*qkvo_proj_count}({num_layers*qkvo_proj_count/TransformerLM_count*100:.2f}%), rope FLOPS: {num_layers*rope_count}({num_layers*rope_count/TransformerLM_count*100:.2f}%)")
```

```python
# d.
configs = {"GPT2_SMALL": {"num_layers": 12, "d_model": 768, "num_heads":12},
           "GPT2_MEDIUM": {"num_layers": 24, "d_model": 1024, "num_heads":16},
           "GPT2_LARGE": {"num_layers": 36, "d_model": 1280, "num_heads":20},
           "GPT2_XL": {"num_layers": 48, "d_model": 1600, "num_heads":25},
           }
for model, config in configs.items():
    num_layers = config["num_layers"]
    d_model = config["d_model"]
    num_heads = config["num_heads"]

    seq_len = context_len
    d_k = d_model // num_heads

    qkvo_proj_count = 8*seq_len*d_model*(num_heads*d_k)
    scaled_dot_count = 4*(seq_len**2)*(num_heads*d_k) + 3*(seq_len**2)*num_heads
    rope_count = 6*seq_len*(num_heads*d_k) 
    MultiheadAttention_count = qkvo_proj_count + scaled_dot_count + rope_count
    ffn_count = 6*seq_len*d_model*d_ff + 6*seq_len*d_ff
    TransformerBlock_count = MultiheadAttention_count + ffn_count + 2*seq_len*d_model 
    TransformerLM_count = num_layers*TransformerBlock_count + 3*seq_len*d_model + 2*seq_len*d_model*vocab_size
    print(f"{model}:Total FLOPS: {TransformerLM_count}, ffn FLOPS: {ffn_count}/layer({num_layers*ffn_count/TransformerLM_count*100:.2f}%), multihead attention FLOPS: {MultiheadAttention_count}/layer({num_layers*MultiheadAttention_count/TransformerLM_count*100:.2f}%), scaled dot product FLOPS: {scaled_dot_count}/layer({num_layers*scaled_dot_count/TransformerLM_count*100:.2f}%), qkvo projection FLOPS: {qkvo_proj_count}/layer({num_layers*qkvo_proj_count/TransformerLM_count*100:.2f}%), rope FLOPS: {rope_count}/layer({num_layers*rope_count/TransformerLM_count*100:.2f}%)")
```
